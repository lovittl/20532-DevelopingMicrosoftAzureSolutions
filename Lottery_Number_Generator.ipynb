```
import random

def generate_lottery_numbers(num_numbers, max_number):
    """
    Generates a list of unique random numbers for a lottery ticket.

    Args:
        num_numbers: The number of unique numbers to generate.
        max_number: The maximum possible number (inclusive).

    Returns:
        A list of integers representing the lottery numbers.
        Returns an empty list if num_numbers is invalid.
    """
    if not isinstance(num_numbers, int) or num_numbers <= 0:
        print("Error: Number of numbers to generate must be a positive integer.")
        return []
    if not isinstance(max_number, int) or max_number <= 0:
        print("Error: Maximum number must be a positive integer.")
        return []
    if num_numbers > max_number:
        print("Error: Number of numbers to generate cannot exceed the maximum number.")
        return []

    numbers = random.sample(range(1, max_number + 1), num_numbers)
    return sorted(numbers) # Return sorted for better readability

def main():
    """
    Main function to run the lottery number generator program.
    """
    print("Welcome to the Lottery Number Generator!")

    # Get user input for the number of numbers and the maximum number.
    while True:
        try:
            num_numbers = int(input("How many numbers do you want to generate for your lottery ticket? "))
            if num_numbers > 0:
                break  # Exit loop if input is valid
            else:
                print("Please enter a positive integer.")
        except ValueError:
            print("Invalid input. Please enter an integer.")
        except EOFError: # added this exception
            print("Error: No input provided. Exiting.")
            return

    while True:
        try:
            max_number = int(input("What is the maximum number for the lottery? "))
            if max_number > 0:
                break  # Exit loop if input is valid
            else:
                print("Please enter a positive integer.")
        except ValueError:
            print("Invalid input. Please enter an integer.")
        except EOFError: # added this exception
            print("Error: No input provided. Exiting.")
            return

    # Generate the lottery numbers.
    lottery_numbers = generate_lottery_numbers(num_numbers, max_number)

    # Print the results.
    if lottery_numbers:  # Check if the list is not empty
        print(f"Your {num_numbers} lucky lottery numbers (from 1 to {max_number}) are:")
        print(lottery_numbers)
        print("Good luck!")
    else:
        print("Failed to generate lottery numbers.  Please check your input and try again.")  # error message

if __name__ == "__main__":
    main()
```
